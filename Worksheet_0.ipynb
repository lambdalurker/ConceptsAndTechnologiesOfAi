```python

time_data = [
    (3.5, 2.0, 7.0), (5.0, 1.5, 6.5), (2.5, 3.0, 8.0),
    (4.0, 2.0, 6.0), (1.5, 4.5, 9.0), (3.0, 2.5, 7.5),
    (5.5, 1.0, 6.0), (2.0, 3.5, 8.5), (4.5, 2.0, 7.0),
    (3.0, 3.0, 7.5), (6.0, 1.5, 6.0), (2.5, 4.0, 8.0),
    (4.0, 2.5, 7.0), (5.0, 2.0, 6.5), (3.5, 2.5, 7.0)
]

print("Number of records:", len(time_data))

low_study = []
moderate_study = []
high_study = []

for day in time_data:
    study_hours = day[0]

    if study_hours < 3:
        low_study.append(study_hours)
    elif study_hours <= 5:
        moderate_study.append(study_hours)
    else:
        high_study.append(study_hours)

print("Low study time:", low_study)
print("Moderate study time:", moderate_study)
print("High study time:", high_study)
```

```python
def sum_nested_list(nested_list):
    """Return the sum of all numeric values in a nested list."""
    total = 0

    for element in nested_list:
        if isinstance(element, list):
            total += sum_nested_list(element)
        else:
            total += element

    return total


nested_list = [1, [2, [3, 4], 5], 6, [7, 8]]
print("Nested list:", nested_list)
print("Total sum:", sum_nested_list(nested_list))

# A second test
print("Second test:", sum_nested_list([1, [2, 3], [4, [5]]]))
```

## Task 2

```python
def generate_permutations(s):
    """Return a list containing every unique permutation of string s."""
    if len(s) <= 1:
        return [s]

    permutations = set()

    for index in range(len(s)):
        chosen_character = s[index]
        remaining_characters = s[:index] + s[index + 1:]

        for smaller_permutation in generate_permutations(remaining_characters):
            permutations.add(chosen_character + smaller_permutation)

    return sorted(permutations)


print('Permutations of "abc":', generate_permutations("abc"))
print('Permutations of "aab":', generate_permutations("aab"))
print('Permutations of an empty string:', generate_permutations(""))
```

```python
directory_structure = {
    "file1.txt": 200,
    "file2.txt": 300,
    "subdir1": {
        "file3.txt": 400,
        "file4.txt": 100
    },
    "subdir2": {
        "subsubdir1": {
            "file5.txt": 250
        },
        "file6.txt": 150
    }
}


def calculate_directory_size(directory):
    """Return the total size in KB of files in a directory and its subdirectories."""
    total_size = 0

    for name, content in directory.items():
        if isinstance(content, dict):
            total_size += calculate_directory_size(content)
        else:
            total_size += content

    return total_size


print("Total directory size:", calculate_directory_size(directory_structure), "KB")
```

```python
def min_coins(coins, amount):
    """Return the minimum number of coins needed, or -1 if the amount is impossible."""
    if amount < 0:
        return -1
    if any(coin <= 0 for coin in coins):
        raise ValueError("Coin denominations must be positive integers.")

    dp = [float("inf")] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for current_amount in range(coin, amount + 1):
            dp[current_amount] = min(
                dp[current_amount],
                dp[current_amount - coin] + 1
            )

    if dp[amount] == float("inf"):
        return -1
    return dp[amount]


print("Coins [1, 2, 5], amount 11:", min_coins([1, 2, 5], 11))
print("Coins [2], amount 3:", min_coins([2], 3))
```

## Task 2

```python
def longest_common_subsequence(s1, s2):
    """Return the length of the longest common subsequence of s1 and s2."""
    rows = len(s1) + 1
    columns = len(s2) + 1
    dp = [[0] * columns for _ in range(rows)]

    for i in range(1, rows):
        for j in range(1, columns):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[len(s1)][len(s2)]


print('LCS length of "abcde" and "ace":',
      longest_common_subsequence("abcde", "ace"))
print('LCS length of "abc" and "def":',
      longest_common_subsequence("abc", "def"))
```

```python
def knapsack(weights, values, capacity):
    """Return the maximum value achievable without exceeding capacity."""
    if len(weights) != len(values):
        raise ValueError("weights and values must have the same length.")
    if capacity < 0:
        raise ValueError("capacity cannot be negative.")
    if any(weight < 0 for weight in weights):
        raise ValueError("item weights cannot be negative.")

    dp = [0] * (capacity + 1)

    for index in range(len(weights)):
        weight = weights[index]
        value = values[index]

        for current_capacity in range(capacity, weight - 1, -1):
            dp[current_capacity] = max(
                dp[current_capacity],
                dp[current_capacity - weight] + value
            )

    return dp[capacity]


weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7

print("Maximum knapsack value:",
      knapsack(weights, values, capacity))
```
