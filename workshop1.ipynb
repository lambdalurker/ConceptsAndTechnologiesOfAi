# Problem - 4: Matrix Operations:

```python
A=np.array([[3,4],[7,8]])
B=np.array([[5,3],[2,1]])
```

1. Prove A.A−1 = I.

```python
A_inv=np.linalg.inv(A)
result= np.dot(A,A_inv)
I=np.eye(2)
print(result)
print(I)
```

2. Prove AB ̸= BA.

```python
AB= np.dot(A,B)
print("AB",AB)
BA=np.dot(B,A)
print("BA",BA)

print("Are AB and BA equal?")
print(np.array_equal(AB, BA))
```

3. Prove \(AB)^T = B^T A^T



```python
AB_transpose = np.transpose(np.dot(A, B))

BTdotAT = np.dot(np.transpose(B), np.transpose(A))

print("(AB) Transpose:")
print(AB_transpose)

print("B Transpose multiplied by A Transpose:")
print(BTdotAT)

print("Are they equal?")
print(np.array_equal(AB_transpose, BTdotAT))
```

Solve the following system of Linear equation using Inverse Methods.

2x − 3y + z = −1

x − y + 2z = −3

3x + y − z = 9

```python
A = np.array([
    [2, -3, 1],
    [1, -1, 2],
    [3, 1, -1]
])

B = np.array([-1, -3, 9])
```

Now for solving we compute A inverse for the equation

**X=A_inv dot B**

```python
A_inv = np.linalg.inv(A)

X = np.dot(A_inv, B)

print("Inverse of A:")
print(A_inv)

print("Solution:")
print(X)

print("Values of x, y and z:")
print("x =", X[0])
print("y =", X[1])
print("z =", X[2])
```

# Experiment: How Fast is Numpy?

1. Element-wise Addition:
• Using Python Lists, perform element-wise addition of two lists of size 1, 000, 000. Measure
and Print the time taken for this operation.

```python
import time
size = 1000000

a = list(range(size))
b = list(range(size))

start = time.time()
result = [a[i] + b[i] for i in range(size)]
end = time.time()

print("Python List Addition Time:", end - start, "seconds")
```

```python
a_np = np.array(a)
b_np = np.array(b)

start = time.time()
result_np = a_np + b_np
end = time.time()

print("NumPy Addition Time:", end - start, "seconds")
```
